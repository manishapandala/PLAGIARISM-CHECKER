## 1. Setup and Data Generation

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from data_generator import generate_covid_dataset, encode_categorical_variables
from descriptive_stats import DescriptiveAnalyzer
from hypothesis_tests import HypothesisTester
from visualizations import Visualizer

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
sns.set_style('whitegrid')
%matplotlib inline

print("✓ Libraries imported successfully")
```

```python
# Generate dataset
print("Generating COVID-19 dataset...")
df = generate_covid_dataset(n_records=10000)
df = encode_categorical_variables(df)
print(f"✓ Dataset generated: {len(df):,} records")
print(f"\nShape: {df.shape}")
print(f"\nColumns: {list(df.columns)}")
```

```python
# Display first few rows
df.head(10)
```

## 2. Descriptive Statistics

```python
# Run descriptive analysis
analyzer = DescriptiveAnalyzer(df)
analyzer.run_full_analysis()
```

```python
# Quick statistical summary
df.describe()
```

## 3. Hypothesis Testing

```python
# Run all hypothesis tests
tester = HypothesisTester(df)
results = tester.run_all_tests()
```

```python
# Display results summary
summary_data = []
for key, result in results.items():
    summary_data.append({
        'Hypothesis': result['hypothesis'],
        'Title': result['title'],
        'P-Value': result['p_value'],
        'Reject Null': result['reject_null'],
        'Test': result['test_used']
    })

results_df = pd.DataFrame(summary_data)
results_df.style.background_gradient(subset=['P-Value'], cmap='RdYlGn_r')
```

## 5. Custom Analysis

Use the cells below for your own exploratory analysis:

```python
# Example: Correlation matrix
numeric_cols = ['HoursWorkedPerDay', 'MeetingsPerDay', 'StressLevel_Numeric', 
                'ProductivityChange_Numeric', 'ProductivityResilienceIndex',
                'TechAdaptation_Numeric', 'CollaborationChallenges_Numeric']

plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Example: Filter data by sector
sector = 'IT'  # Change to: IT, Healthcare, Retail, Education, Finance
sector_df = df[df['Sector'] == sector]
print(f"\n{sector} Sector Statistics (n={len(sector_df):,}):")
print(f"Remote Work: {sector_df['WorkFromHome'].mean()*100:.1f}%")
print(f"Avg Stress: {sector_df['StressLevel_Numeric'].mean():.2f}")
print(f"Avg Productivity: {sector_df['ProductivityChange_Numeric'].mean():.2f}")
print(f"Avg PRI: {sector_df['ProductivityResilienceIndex'].mean():.1f}")
```
